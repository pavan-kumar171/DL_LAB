```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Dense, Flatten

corpus=[
    "I like deep learning",
    "Deep learning models can learn word embeddings",
    "word embeddings capture semantic meaning of words",
    "Neural networks are powerful for NLP tasks",
    "Deep learning is connected to Machine learning",
    "I like Machine learning"
]
tokenizer=Tokenizer()
tokenizer.fit_on_texts(corpus)
word_index=tokenizer.word_index
print("Vocabulary:",word_index)
sequences=tokenizer.texts_to_sequences(corpus)
padded_sequences=pad_sequences(sequences, padding='post')
print("\nSequences:\n",sequences)
print("\nPadded Sequences:\n",padded_sequences)
vocab_size=len(word_index)+1
embedding_dim=8
max_length=padded_sequences.shape[1]
model=Sequential([
    Embedding(input_dim=vocab_size,output_dim=embedding_dim,input_length=max_length),
    Flatten(),
    Dense(10,activation='relu'),
    Dense(5,activation='relu'),
    Dense(1,activation='sigmoid')
])
model.compile(optimizer='adam',loss='binary_crossentropy')
labels=np.array([1,0,1,0,1,0])
model.fit(padded_sequences,labels,epochs=100,verbose=0)
embeddings=model.layers[0].get_weights()[0]
print("\nWord Embeddings shape:",embeddings.shape)
word="deep"
word_id=word_index[word]
print(f"Embeddings for word '{word}':\n", embeddings[word_id])
word="learning"
word_id=word_index[word]
print(f"Embeddings for word '{word}':\n", embeddings[word_id])
word="like"
word_id=word_index[word]
print(f"Embeddings for word '{word}':\n", embeddings[word_id])
word="machine"
word_id=word_index[word]
print(f"Embeddings for word '{word}':\n", embeddings[word_id])
```
